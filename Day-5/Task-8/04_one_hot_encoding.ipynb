**Libraries**

````python
import pandas as pd
````

**Load Dataset and View**

````python
df=pd.read_csv("/content/ai4i2020.csv")
df['Type'].head()
````

**One Hot Encoding**

````python
dummies=pd.get_dummies(df['Type'])
dummies.head()
````

**Merge dummy columns**

````python
merged=pd.concat([df,dummies],axis='columns')
merged.head()
````

**Drop unecessary columns**

````python
final=merged.drop('Type',axis='columns')
````

**Drop dummy column**

````python
final=final.drop('M',axis='columns')
final.head()
````

**All these above steps can be done in single step:**
(one hot encoding, merge, drop unecessary columns)
```
dummies=pd.get_dummies(df,columns=['Type'],drop_first=True)
```

*drop_first=True*  **is to avoid dummy trap**

**Save the clean dataset**

````python
final.to_csv('encoded_factory_data.csv', index=False)
````

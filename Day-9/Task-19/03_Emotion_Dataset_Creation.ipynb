```python
import cv2
import os
from ultralytics import YOLO
```

```python
model=YOLO('yolov8n.pt')
```

```python
emotion_map = {
    "01": "neutral",
    "02": "calm",
    "03": "happy",
    "04": "sad",
    "05": "angry",
    "06": "fearful",
    "07": "disgust",
    "08": "surprised"
}
```

```python
for emotion in emotion_map.values():
    os.makedirs(
        f'/content/emotion_dataset/{emotion}',
        exist_ok=True
    )
```

```python
video_folder = '/content/videos'
videos = os.listdir(video_folder)
print(len(videos))
```

```python
count = 0
for video in videos:
    video_path = os.path.join(
        video_folder,
        video
    )
    filename = os.path.basename(video_path)
    emotion_code = filename.split('-')[2]
    emotion = emotion_map[emotion_code]
    cap = cv2.VideoCapture(video_path)
    ret, frame = cap.read()
    if not ret:
        continue
    results = model(frame)
    boxes = results[0].boxes
    for box in boxes:
        coords = box.xyxy[0]
        x1, y1, x2, y2 = map(int, coords)
        cropped = frame[y1:y2, x1:x2]
        cropped = cv2.resize(
            cropped,
            (224,224)
        )
        save_path = f'/content/emotion_dataset/{emotion}/{count}.jpg'
        cv2.imwrite(
            save_path,
            cropped
        )
        count += 1
    cap.release()
```

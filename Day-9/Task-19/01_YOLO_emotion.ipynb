```python
import cv2
import os
import matplotlib.pyplot as plt
from ultralytics import YOLO
```

```python
model = YOLO('yolov8n.pt')
```

```python
emotion_map = {
    "01": "Neutral",
    "02": "Calm",
    "03": "Happy",
    "04": "Sad",
    "05": "Angry",
    "06": "Fearful",
    "07": "Disgust",
    "08": "Surprised"
}
```

```python
video_path = '/content/01-01-04-01-01-01-01.mp4'
```

```python
filename = os.path.basename(video_path)
emotion_code = filename.split('-')[2]
emotion = emotion_map[emotion_code]
print(emotion)
```

```python
cap = cv2.VideoCapture(video_path)
```

```python
while True:
    ret, frame = cap.read()
    if not ret:
        break

    results = model(frame)
    annotated_frame = results[0].plot()
    cv2.putText(
        annotated_frame,
        f'Emotion: {emotion}',
        (20, 40),
        cv2.FONT_HERSHEY_SIMPLEX,
        1,
        (0,255,0),
        2
    )
    plt.figure(figsize=(8,8))
    plt.imshow(
        cv2.cvtColor(
            annotated_frame,
            cv2.COLOR_BGR2RGB
        )
    )
    plt.axis('off')
    plt.show()
```

```python
results = model(frame)
```

```python
result_frame = results[0].plot()
```

```python
cap.release()
```

```python
cv2.imwrite(
    'output.jpg',
    annotated_frame
)
```
